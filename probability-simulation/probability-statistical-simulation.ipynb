# Probability & Statistical Simulation with NumPy

This portfolio notebook demonstrates how Python and NumPy can be used to approximate probabilities through repeated simulation.

## Skills Demonstrated

- Random number generation with NumPy
- Reproducible simulations using a random seed
- Boolean conditions and array operations
- Monte Carlo-style probability estimation
- Comparing simulation results with probability concepts

> **Note:** This notebook is a portfolio adaptation of coursework completed while learning Python and NumPy.

```python
import numpy as np

np.random.seed(143)
```

## Rolling a 6 or 8 with Two Dice

Two standard six-sided dice are rolled 10,000 times. The simulation estimates the probability that their sum is either 6 or 8.

```python
trials = 10_000

die1 = np.random.randint(1, 7, trials)
die2 = np.random.randint(1, 7, trials)

totals = die1 + die2
prob_6_or_8 = np.mean((totals == 6) | (totals == 8))

print(f"Estimated probability: {prob_6_or_8:.4f}")
```

### Interpretation

The simulation uses many repeated trials to estimate the long-run probability of rolling a total of 6 or 8.

## Rolling a 7 or 11 with Two Dice

This simulation estimates the probability of rolling a total of 7 or 11 with two standard dice.

```python
die1 = np.random.randint(1, 7, trials)
die2 = np.random.randint(1, 7, trials)

totals = die1 + die2
prob_7_or_11 = np.mean((totals == 7) | (totals == 11))

print(f"Estimated probability: {prob_7_or_11:.4f}")
```

## Three Dice with a Sum of 7 or Less

Three standard dice are rolled 10,000 times. The simulation estimates the probability that the total is 7 or less.

```python
dice = np.random.randint(1, 7, size=(trials, 3))
totals = dice.sum(axis=1)

prob_sum_7_or_less = np.mean(totals <= 7)

print(f"Estimated probability: {prob_sum_7_or_less:.4f}")
```

## Twenty Coin Flips

This experiment simulates 20 fair coin flips across 10,000 trials. It estimates the probability of getting fewer than 7 heads or more than 13 heads.

```python
coin_flips = np.random.randint(0, 2, size=(trials, 20))
heads = coin_flips.sum(axis=1)

prob_extreme_heads = np.mean((heads < 7) | (heads > 13))

print(f"Estimated probability: {prob_extreme_heads:.4f}")
```

### Interpretation

Boolean conditions make it possible to count outcomes that satisfy more than one rule, such as values in either tail of the distribution.

## Standard Normal Simulation

One million values are generated from a standard normal distribution. The simulation estimates how often observations fall more than 3 standard deviations above the mean.

```python
normal_values = np.random.standard_normal(1_000_000)

count_above_3 = np.sum(normal_values > 3)
prob_above_3 = np.mean(normal_values > 3)

print(f"Count above 3 standard deviations: {count_above_3}")
print(f"Estimated probability: {prob_above_3:.5f}")
```

## Summary

These simulations show how NumPy arrays and repeated random trials can be used to approximate probabilities efficiently. The same workflow can be extended to larger statistical simulations and uncertainty analysis.
