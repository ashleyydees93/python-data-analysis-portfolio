# Data Visualization with Matplotlib

This portfolio notebook demonstrates data visualization techniques developed while learning Python, NumPy, and Matplotlib. The examples focus on plotting mathematical relationships, comparing functions, visualizing two-dimensional data, and organizing multiple plots into a clear figure.

## Skills Demonstrated

- Creating line and scatter plots
- Generating plotting data with NumPy
- Comparing multiple functions on the same axes
- Using labels, legends, titles, and axis limits
- Building multi-panel figures
- Adding text and annotations to visualizations

> **Note:** This notebook is a portfolio adaptation of coursework completed while learning Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Comparing Two Functions

The first visualization compares two functions over the interval $[-2, 5]$:

- $y = 3x^2 - 2$
- $y = 30 - x^2$

Plotting both functions on the same axes makes it easy to compare their shapes and identify where their values are similar or different.

```python
x = np.linspace(-2, 5, 200)

y1 = 3 * x**2 - 2
y2 = 30 - x**2

plt.figure(figsize=(8, 5))
plt.plot(x, y1, label='y = 3x² - 2')
plt.plot(x, y2, label='y = 30 - x²')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Comparison of Two Quadratic Functions')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

### Key Observation

The two curves have different orientations and rates of change. Displaying them together makes their relationship and intersection behavior easier to interpret visually.

## Polynomial and Linear Function Comparison

This visualization compares a cubic polynomial with a linear function:

- $y = x^3 - 3x^2 + 3$
- $y = 2x$

This example demonstrates how multiple mathematical relationships can be displayed together with clear labels and a legend.

```python
x = np.linspace(-2, 5, 300)

polynomial = x**3 - 3*x**2 + 3
linear = 2*x

plt.figure(figsize=(8, 5))
plt.plot(x, polynomial, label='y = x³ - 3x² + 3')
plt.plot(x, linear, label='y = 2x')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Polynomial vs. Linear Function')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

### Key Observation

The linear function changes at a constant rate, while the cubic polynomial changes direction and grows at a nonconstant rate. Plotting them together highlights these differences clearly.

## Two-Dimensional Function Visualization

This example uses randomly generated points in a two-dimensional region and evaluates the function

$$f(x,y) = (x-3)^2 - (y-3)^2.$$

The function values are used to organize the scatter plot, demonstrating how numerical values can be mapped to points in two-dimensional space.

```python
np.random.seed(143)

n = 300
x = np.random.uniform(-2, 2, n)
y = np.random.uniform(-2, 2, n)

values = (x - 3)**2 - (y - 3)**2

plt.figure(figsize=(7, 6))
scatter = plt.scatter(x, y, c=values)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Two-Dimensional Function Values')
plt.colorbar(scatter, label='f(x, y)')
plt.show()
```

### Key Observation

The scatter plot combines point location with a third numerical quantity, making it possible to see how the function changes across the two-dimensional region.

## Multi-Panel Visualization

Subplots are useful when several related views need to be presented in one figure. This example creates three axes with `subplot2grid()` and displays related functions in separate panels.

```python
x = np.linspace(-2.5, 2.5, 200)

fig = plt.figure(figsize=(10, 10))

ax1 = plt.subplot2grid((2, 2), (0, 0), colspan=2)
ax2 = plt.subplot2grid((2, 2), (1, 0))
ax3 = plt.subplot2grid((2, 2), (1, 1))

ax1.plot(x, x**2)
ax1.set_title('Quadratic Function')
ax1.set_xlabel('x')
ax1.set_ylabel('y')

ax2.plot(x, x**3)
ax2.set_title('Cubic Function')
ax2.set_xlabel('x')
ax2.set_ylabel('y')

ax3.plot(x, np.abs(x))
ax3.set_title('Absolute Value Function')
ax3.set_xlabel('x')
ax3.set_ylabel('y')

fig.suptitle('Multi-Panel Function Visualization', fontsize=16)
plt.tight_layout()
plt.show()
```

### Key Observation

Using multiple axes allows related visualizations to be compared without overcrowding a single plot. This layout approach is useful for dashboards, exploratory analysis, and side-by-side comparisons.

## Annotation and Axis Control

The final example demonstrates axis limits and plot annotation. These features help direct attention to important locations or values in a visualization.

```python
x = np.linspace(-2.5, 2.5, 200)
y = x**2

fig, ax = plt.subplots(figsize=(7, 6))
ax.plot(x, y)

ax.set_xlim(-2.5, 2.5)
ax.set_ylim(-0.5, 5.5)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Annotated Quadratic Function')

ax.annotate(
    'Minimum',
    xy=(0, 0),
    xytext=(0.7, 2),
    arrowprops={'arrowstyle': '->'}
)

plt.show()
```
