## Factorial Function

The factorial of a positive integer is the product of that integer and every positive integer below it. This implementation uses a `while` loop.

```python
def factorial(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n must be a non-negative integer")

    result = 1
    while n > 1:
        result *= n
        n -= 1

    return result

print(factorial(5))
```

## Positive Divisors

The `div(n)` function returns every positive divisor of a positive integer and validates the input before performing the calculation.

```python
def div(n):
    if not isinstance(n, int):
        raise TypeError("n must be an integer")

    if n <= 0:
        raise ValueError("n must be positive")

    divisors = []

    for i in range(1, n + 1):
        if n % i == 0:
            divisors.append(i)

    return divisors

print(div(12))
```

## Numerical Derivative Approximation

A derivative can be approximated numerically by measuring how much a function changes over a very small interval.

```python
def deriv(f, a, h=0.001):
    return (f(a + h) - f(a)) / h

def f1(x):
    return x**2 - 3*x + 4

def f2(x):
    return x**3 + x**2 - 3*x

print(f"Approximate derivative of f1 at x=5: {deriv(f1, 5):.3f}")
print(f"Approximate derivative of f2 at x=1: {deriv(f2, 1):.3f}")
```

## NumPy Array Indexing

NumPy arrays allow data to be organized into rows and columns and selected efficiently by position.

```python
wa = np.arange(20).reshape(4, 5)

print("Array:")
print(wa)

print("\nSecond row:")
print(wa[1])

print("\nThird column:")
print(wa[:, 2])

print("\nBottom-right value:")
print(wa[-1, -1])
```

## Boolean Masks

Boolean masks provide a concise way to filter an array using logical conditions.

```python
mask = (wa < 5) | (wa > 16)

selected_values = wa[mask]

print("Values less than 5 or greater than 16:")
print(selected_values)
```
